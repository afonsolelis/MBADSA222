## Glossário

- **Default**: inadimplência maior que 90 dias
- **Árvore de decisão**: algoritmo de machine learning que tem como objetivo classificar uma variável resposta de acordo com um conjunto de variáveis (*features*)
- **Ramo**: uma regra de quebra na árvore de decisão (ex: possui veículo? SIM e NÃO vão definir dois sub-públicos)
- **Folha**: A última "caixinha" da árvore de decisão
- ***Overfitting***: Om fenômeno que ocorre quando o modelo fica muito bom na base de treinamento mas seu desempenho cai na base de validação (ou qualquer outra base)
- **Base de desenvolvimento**: Base utilizada no desenvolvimento do algoritmo
- **Base de validação** *out-of-sample*: amostra da base de dados disponível que não é utilizada no desenvolvimento do modelo, mas sim para avaliar a sua generalização. Pode ser utilizada para ajustes no modelo.
- **Base de validação *out-of-time***: uma base que não é utilizada para desenvolver nem para ajustar o modelo, apenas para a avaliação final - consiste em um período mais recente que a base de desenvolvimento.
- **Poda de uma árvore**: a eliminação de folhas visando melhorar a generalização do algoritmo.
- **Pré-poda**: uma poda realizada sem a observação de uma base de validação.
- **Pós-poda**: uma poda realizada com base em métricas de desempenho da base de validação.
- **Escore de crédito**: um número advindo de um modelo (regressão, árvore ou outro) que classifica a probabilidade de inadimplência. Em geral é uma *probabilidade de não default* multiplicada por 1000, sem decimais.
- ***Ensemble***: Método de mistura de modelos que consiste em extrair a média das preditivas de dois ou mais modelos buscando um modelo ainda melhor que os anteriores.
- ***Bootstrapping***: Metodologia de estimação de parâmetros baseada em reamostragens sucessivas da mesma amostra.
- ***Bagging***: Abreviação de ***ootstrap aggregation***. Consiste em construir modelos sequencialmete sobre amostras *bootstrap*, e agregar os modelos através da média de suas preditivas.
- ***Boosting*** Construção sequencial de modelos, em que um modelo tenta reduzir o erro do anterior.
- ***Gradient boosting***: Método popular de boosting que consiste na construção sequencial de árvores de decisão em que cada um tem como variável resposta o resíduo do anterior.
- ***Extreme gradient boosting***: Implementação do *gradient boosting* otimizada, que se tornou popular por ser utilizada em soluções vencedoras de competições de ciência de dados.

### 1. Ajustando uma árvore de decisão. <a name="1"></a>
[Índice](#topo)

Vamos ajustar uma árvore para prever ```mau```, utilizando algumas variáveis para refrescar a nossa memória.

Aqui já estamos construindo as bases na forma como precisamos usando o *patsy* nas três bases (DES, OOS e OOT).

A árvore propriamente dita é construída com a função ```DecisionTreeClassifier()``` e guardada no objeto ```clf```.

Em seguida, estamos calculando a probabilidade de "mau" (PD) e a classificação simples indicada pelo algoritmo.

````python
%%time
# O pacote patsy permite ajustar um modelo com uma equação, como no R
equacao = '''mau ~ sexo + posse_de_veiculo + posse_de_imovel + qtd_filhos + tipo_renda + educacao 
    + estado_civil + tipo_residencia + idade + qt_pessoas_residencia'''
           
# A função é a dmatrices, do patsy. Na prática, essa função monta a matriz de modelagem
y_des, x_des = patsy.dmatrices(equacao, data=des, return_type='dataframe')
````

````python
x_des
````

````python
%%time
equacao = '''mau ~ sexo + posse_de_veiculo + posse_de_imovel + qtd_filhos + tipo_renda + educacao 
    + estado_civil + tipo_residencia + idade + qt_pessoas_residencia'''
           
y_des, x_des = patsy.dmatrices(equacao, data=des)
y_oot, x_oot = patsy.dmatrices(equacao, data=oot)
y_oos, x_oos = patsy.dmatrices(equacao, data=oos)


# Árvore
clf = DecisionTreeClassifier().fit(x_des,y_des)

des['classe_predita'] = clf.predict(x_des)
des['PD'] = clf.predict_proba(x_des)[:,1]

oos['classe_predita'] = clf.predict(x_oos)
oos['PD'] = clf.predict_proba(x_oos)[:,1]

oot['classe_predita'] = clf.predict(x_oot)
oot['PD'] = clf.predict_proba(x_oot)[:,1]
````

#### Desempenho da árvore

Vamos avaliar o desempenho da árvore na base de desenvolvimento e na base de validação (```des```,  e ```oos```, respectivamente). Vamos avaliar os índices KS, Gini, AUC e acurácia.

#### Função para calcular o desempenho

Para facilitar essas comparações, vamos construir uma função para calcular todos esses indicadores. É interessante colocar valores padrão para os parâmetros da função, assim, vamos poder indicar somente a base de dados para rodar a função com os valores padrão dela.

````python
# Calcular o Gini
def calcula_gini(RESP, PD):
    #AUC
    auc = roc_auc_score(RESP, PD)
    
    #Gini
    gini = 2*auc -1
    return gini, auc


def desempenho_df(dados = des
               , PD = 'PD'
               , CLASSE_PRED = 'classe_predita'
               , RESP = 'mau'):
    
#     Calculando a acuracia
    acc = accuracy_score(dados[RESP]
                                 , dados[CLASSE_PRED])

    #AUC
    #Gini
    gini, auc = calcula_gini(dados[RESP], dados[PD])
    
    #KS
    ks = ks_2samp(dados.loc[dados[RESP] == 1, PD]
                  , dados.loc[dados[RESP] !=1, PD]).statistic

    print('Acurácia: {0:.2f}%'.format(acc*100))
    print('AUC: {0:.2f}%'.format(auc*100))
    print('GINI: {0:.2f}%'.format(gini*100))
    print('KS: {0:.2f}%\n'
          .format(ks*100))
    return None
````

#### Desempenho em desenvolvimento e OOS

Co a árvore construida e a função que avalia o desempenho feita, vamos avaliá-la.

````python
desempenho_df(dados = des)
desempenho_df(dados = oos)

calcula_gini(des['mau'], des['PD'])
````

**OBS**: a árvore perde muito desempenho da base de desenvolvimento para validação. De fato é um algoritmo que propicia muito *overfitting* se não tomarmos cuidado.

#### Poda

Visto o desempenho na base de validação, vamos tentar buscar uma árvore mais parcimoniosa, que possa até ter menos desempenho na base de desenvolvimento, mas buscaremos uma que tenha melhor desempenho na base de validação.

Vamos alterar os seguintes parâmetros:

**Número mínimo de observações por folha** (```min_samples_leaf```)   <br>
**Profundidade máxima da árvore** (```max_depth```)



````python
%%time

clf = DecisionTreeClassifier(min_samples_leaf=10, max_depth=8).fit(x_des,y_des)

des['classe_predita'] = clf.predict(x_des)
des['PD'] = clf.predict_proba(x_des)[:,1]

oos['classe_predita'] = clf.predict(x_oos)
oos['PD'] = clf.predict_proba(x_oos)[:,1]

oot['classe_predita'] = clf.predict(x_oot)
oot['PD'] = clf.predict_proba(x_oot)[:,1]

desempenho_df(dados = des)
desempenho_df(dados = oos)
````

**OBS:** A árvore apresenta um desempenho bem melhor na base "OOT".

### Otimizando a árvore: *Grid search*

O ***grid search*** (busca em árvore) é um método popular em *machine learning* para a otimização de hiperparâmetros de um algoritmo, que consiste em definir um conjunto de valores para os hiperparâmetros, avaliar o resultado em todas as combinações, e escolher a combinação que fornece o melhor desempenho na ```oos```.

No exemplo abaixo, vamos avaliar 20 possibilidades de número mínimo de observações por folha e 20 possibilidades de profundidade máxima da árvore, totalizando 400 combinações - isso aí, vamos avaliar 400 árvores!

#### GridSearchCV

Com essa função, precisaremos indicar o modelo, e para cada parâmetro que quisermos indicar com uma lista, a função vai fazer o *fit* do modelo com todas as combinações de parâemtros das listas indicadas, e vai calcular a métrica especificada para comparar os modelos.

**Dicionário de parâmetros**: Vamos usar usar uma forma interessante de inserir os parâmetros (ou as combinações deles) no *gridsearch*: vamos construir um dicionário. Sempre que o valor for uma lista, todas as possibilidades serão testadas. Se houver duas ou mais listas, o algoritmo monta a *grade* combinando as duas (ou mais) listas.

**parâmetro CV:** Significa *cross validation* (ou validação cruzada). Nele indicamos os índices das observações que serão utilizadas para desenvolver o modelo e para validar respectivamente. Através deste parâmetro, a função ainda consegue fazer outras modalidades de validação cruzada.

````python
y_prd, x_prd = patsy.dmatrices(equacao, data=periodo_des)
````

````python
list(range(1,100,5))
````

````python
list(range(1,20,1))
````

````python
%%time
from sklearn.model_selection import GridSearchCV

indice_des = periodo_des[bool_des].index.values
indice_oos = periodo_des[~bool_des].index.values

clf = DecisionTreeClassifier()

parametros = { 'min_samples_leaf': list(range(1,100,5))
              , 'max_depth': list(range(1,20,1))}

grid = GridSearchCV(estimator = clf,         
                    param_grid = parametros, 
                    scoring = 'roc_auc',   
                    verbose=False,
                    cv=[[indice_des.tolist(), indice_oos.tolist()]]
                   )                 

grid.fit(x_prd.base,y_prd.base.flatten())
resultados_grid = pd.DataFrame(grid.cv_results_)
````

A função retorna uma tabela de resultados, com uma linha para cada modelo testado, com suas características, tempo de treinamento, métrica de desempenho e o *ranking* do modelo.

### 2. *Random Forest*<a name="2"></a>
[Índice](#topo)

O algoritmo do Random Forest é o mesmo do *bootstrap*, com uma pequena modificação: além de amostrar as observações, amostramos também as colunas, de modo a gerar árvores mais diferentes.

Temos uma amostra de treinamento (o mesmo que desenvolvimento) de tamanho ```N```, com ```P``` variáveis, escolhemos um número M de árvores para se treinar e seguimos os seguintes passos:

1. Para cada $j$ variando de 1 a M ($j=1, 2,... M$):
    1. amostrar com reposição N indivíduos e ```p``` variáveis com ```p<P```
    2. Treinar uma árvore de decisão $A_j$

Esse basicamente é o nosso modelo: um conjunto de M árvores geradas com uma certa aleatoriedade.

Podem restar agora duas perguntas:

1) Como classificamos um indivúduo em M árvores?
2) Como obtemos a probabilidade de esse indivíduo novo pertenter a cada classe (default e não default no nosso caso)?

As duas perguntas se respondem de forma relatiamente simples: da mesma forma que o *bootstrappint*, através de uma média.

Ou seja, para obter a probabilidade de classificação em cada classe fazemos o seguinte:
- "Escoramos" o indivíduo nas M árvores
- Tiramos uma média das probabilidades de classificação 
- A maior probabilidade define a classificação desse indivíduo

A idéia é muito simples não é mesmo?

````python
%%time
from sklearn.ensemble import RandomForestClassifier

clf = RandomForestClassifier(n_estimators=600, max_depth=None
                             , min_samples_leaf=1).fit(x_des,y_des.flatten())

des['classe_predita'] = clf.predict(x_des)
des['PD'] = clf.predict_proba(x_des)[:,1]

oos['classe_predita'] = clf.predict(x_oos)
oos['PD'] = clf.predict_proba(x_oos)[:,1]

oot['classe_predita'] = clf.predict(x_oot)
oot['PD'] = clf.predict_proba(x_oot)[:,1]
````

### 2.1. Hiperparâmetros<a name="2.1"></a>
[Índice](#topo)

Hiperparâmetros são parâmetros que controlam como o algoritmo é otimizado. O número de árvores por exemplo é um hiperparâmetro da *random forest*.

Qual é o melhor conjunto de hiperparâmetros para o *random forest*? Esta é a pergunta de ouro :) Mas felizmente há um método estruturado para se buscar esses hiperparâmetros, esse método é conhecido como *grid search* (ou busca por grade).

Vamos exemplificar essa busca de modo simplificado abaixo, buscando parâmetros um a um, em seguida variando mais de um ao longo de uma *grade* como o nome do método sugere.


#### <span style="color:red">AVISO! não rodar este :) </span>
Esse tipo de método pode ser um pouco custoso para a máquina, receio que diversas pessoas rodando na mesma máquina pode ser um tanto... pesado.

#### Gridsearch 1 - variando o número de árvores

Vamos estabelecer parâmetros fixos de ```max_depth``` e ```min_samples_leaf```, e variar o número de árvores, e avaliar a cada rodada o desempenho do modelo.

````python
%%time
lista_gini_des=[]
lista_gini_oos=[]


grid = list(range(1, 1000, 20))

for num_arvores in grid:
    clf = RandomForestClassifier(n_estimators=num_arvores
                                 , max_depth=10
                                 , min_samples_leaf=5).fit(x_des,y_des.flatten())
    
    gini_des, _ = calcula_gini(y_des, clf.predict_proba(x_des)[:,1])
    gini_oos, _ = calcula_gini(y_oos, clf.predict_proba(x_oos)[:,1])
    
    lista_gini_des.append(gini_des)
    lista_gini_oos.append(gini_oos)
````

````python
fig, ax = plt.subplots()
ax.plot(lista_gini_des, '-b', label='DES')
ax.plot(lista_gini_oos, '--r', label='OOS')

leg = ax.legend();
ticks = ax.set_xticks(range(0, len(grid), 10))
labels = ax.set_xticklabels(labels=grid[::10], rotation=30, fontsize='small')
````

#### Gridsearch 2 - variando mínimo de observações por folha

Agora vamos variar apenas ```min_samples_leaf```.

````python
%%time
lista_gini_des=[]
lista_gini_oos=[]

grid=list(range(2,5000,100))

for n_min_obs in grid:
    clf = RandomForestClassifier(n_estimators=30
                                 , max_depth=10
                                 , min_samples_leaf=n_min_obs )\
          .fit(x_des,y_des.flatten())
    gini_des, _ = calcula_gini(y_des, clf.predict_proba(x_des)[:,1])
    gini_oos, _ = calcula_gini(y_oos, clf.predict_proba(x_oos)[:,1])
    
    lista_gini_des.append(gini_des)
    lista_gini_oos.append(gini_oos)
````

````python
fig, ax = plt.subplots()
ax.plot(lista_gini_des, '-b', label='DES')
ax.plot(lista_gini_oos, '--r', label='OOS')

leg = ax.legend();
ticks = ax.set_xticks(range(0, len(grid), 5))
labels = ax.set_xticklabels(labels=grid[::5], rotation=30, fontsize='small')
ax.set_title("Random Forest - Gini vs N mínimo por folha")
ax.set_xlabel("Número mínimo de observações por folha")
ax.set_ylabel("Gini");
````

#### Gridsearch 3  - variando a profundidade da árvore

Agora vamos fixar ```n_estimators``` e ```min_samples_leaf``` e variar ```max_depth```.

````python
%%time
lista_gini_des=[]
lista_gini_oos=[]

grid=list(range(1,50))
for profundidade in grid:
    clf = RandomForestClassifier(n_estimators=30
                                 , max_depth=profundidade
                                 , min_samples_leaf=1 
                                ).fit(x_des,y_des.base.flatten())
    gini_des, _ = calcula_gini(y_des, clf.predict_proba(x_des)[:,1])
    gini_oos, _ = calcula_gini(y_oos, clf.predict_proba(x_oos)[:,1])
    
    lista_gini_des.append(gini_des)
    lista_gini_oos.append(gini_oos)
````

````python
fig, ax = plt.subplots()
ax.plot(lista_gini_des, '-b', label='DES')
ax.plot(lista_gini_oos, '--r', label='OOS')

leg = ax.legend();
ticks = ax.set_xticks(range(0, len(grid), 10))
labels = ax.set_xticklabels(labels=grid[::10], rotation=30, fontsize='small')
ax.set_title("Random Forest - Gini vs Profundidade máxima")
ax.set_xlabel("Profundidade máxima")
ax.set_ylabel("Gini")
````

#### Gridsearch 4 - alterar profundidade e número de observações por folha

````python
%%time
grid_dic = {'obs_folha':[], 'profund_max':[], 'gini_des':[], 'gini_oos': []}

grid_profund=list(range(3,20,2))
grid_obs_folha=list(range(1,30,1))

for profundidade in grid_profund:
    for obs_folha in grid_obs_folha:
        clf = RandomForestClassifier(n_estimators=30
                                     , max_depth=profundidade
                                     , min_samples_leaf=obs_folha 
                                    ).fit(x_des,y_des.base.flatten())
        gini_oos, roc = calcula_gini(y_oos.base.flatten(), clf.predict_proba(x_oos)[:,1])
        gini_des, roc = calcula_gini(y_des.base.flatten(), clf.predict_proba(x_des)[:,1])

        grid_dic['gini_oos'].append(gini_oos)
        grid_dic['gini_des'].append(gini_des)

        grid_dic['obs_folha'].append(obs_folha)
        grid_dic['profund_max'].append(profundidade)

grid=pd.DataFrame(grid_dic)
````

````python
sns.heatmap(pd.DataFrame(grid_dic).pivot_table(index='profund_max'
                                   , columns='obs_folha'
                                   , values='gini_oos'
                                   , aggfunc='mean'));
# Aurélien Géron
````

#### Conclusão

O *Random Forest* é um modelo mais robusto, mas também mais pesado que a árvore. As buscas são mais penosas, mas o tratamento de *overfitting* é bem mais robusto.

### 3. *Gradient Boosting*<a name="3"></a>
[Índice](#topo)

***Boosting:*** é qualquer método de *ensemble* (mistura de modelos) que tenha a finalidade de combinar modelos *fracos* de modo a resultar em um *forte*. A forma mais popular do *boosting* hoje consiste em uma construção sequencial de modelos, cada um tentando reduzir o erro do anterior. Para se obter a preditiva, basta encadear novamente as previsões de cada modelo somando-as.

***Gradient Boosting\:*** É o algoritmo de *boosting* mais popular. Consiste em construir árvores sequencialmente, em que uma árvore tenta *melhorar* a previsão sobre o erro da anterior. A lógica do algoritmo é a seguinte:

1. Ajustar uma árvore de decisão na base de desenvolvimento
2. Gravar os *desvios* de previsão em cada observação (resíduos)
3. Ajustar uma nova árvore para explicar os resíduos obtidos em 2
4. Repetir 1, 2 e 3 até atingir um critério de parada

***XGBoosting\:*** (de *extreme gradient boosting*) é uma implementação do *gradient boosting* bastante eficiente e de código aberto desenvolvido pela Tianqi Chen, que ficou bastante popular por ser utilizada em diversas implementações de soluções vencedoras em competições de ciência de dados pelo mundo todo. #### Rodando o *gradient boosting*

Vamos rodar abaixo o *gradient boosting* pelo scikitlearn.

Comparando com *Random Forest*, este algoritmo é bem mais delicado, com mais hiperparâmetros, e necessita de mais cuidados para se evitar o *overfitting*. Adicionalmente aos parâmetros que já vimos, temos:

***learning_rate***: este parâmetro tem o intuito de reduzir a contribuição de cada árvore (quando menor que 1). Com contribuições menores e mais árvores, o algoritmo deve captar padrões mais sutis. Há um equilíbrio entre o ```learning_rat``` e o ```n_estimators```.

````python
from sklearn.ensemble import GradientBoostingClassifier
````

````python
%%time
parametros = {'n_estimators': 100
              , 'max_depth': None
              , 'min_samples_leaf': 5
              , 'learning_rate': .1
             }

clf = GradientBoostingClassifier(**parametros)\
        .fit(x_des,y_des.base.flatten())

````

#### Desempenho do GB

````python
des['classe_p_gb'] = clf.predict(x_des)
des['PD_gb'] = clf.predict_proba(x_des)[:,1]

oos['classe_p_gb'] = clf.predict(x_oos)
oos['PD_gb'] = clf.predict_proba(x_oos)[:,1]

oot['classe_p_gb'] = clf.predict(x_oot)
oot['PD_gb'] = clf.predict_proba(x_oot)[:,1]

desempenho_df(dados = des, PD = 'PD_gb', CLASSE_PRED = 'classe_p_gb')
desempenho_df(dados = oos, PD = 'PD_gb', CLASSE_PRED = 'classe_p_gb')
````
